```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

wine = load_wine(as_frame=True)

df = wine.frame.copy()

print(df.shape)
display(df.head())

print("Jumlah data:", df.shape[0])
print("Jumlah fitur:", len(wine.feature_names))
```

T/E/P Dataset Wine

T (Task)
Tugas model adalah memprediksi kelas wine berdasarkan karakteristik kimianya.

E (Experience)
Model belajar menggunakan data wine yang sudah memiliki label kelas.

P (Performance)
Performa model dilihat menggunakan accuracy dan classification report pada data test.

```python
df['target'].value_counts().plot(kind='bar')

plt.title("Jumlah Data Setiap Kelas Wine")
plt.xlabel("Kelas")
plt.ylabel("Jumlah Data")

plt.show()


plt.hist(df['alcohol'])

plt.title("Distribusi Alcohol")
plt.xlabel("Alcohol")
plt.ylabel("Frekuensi")

plt.show()

for kelas in df['target'].unique():

    data = df[df['target'] == kelas]

    plt.scatter(
        data['alcohol'],
        data['color_intensity'],
        label=f'Kelas {kelas}'
    )

plt.title("Alcohol vs Color Intensity")
plt.xlabel("Alcohol")
plt.ylabel("Color Intensity")
plt.legend()

plt.show()

plt.boxplot([
    df[df['target'] == kelas]['proline']
    for kelas in df['target'].unique()
])

plt.title("Distribusi Proline Setiap Kelas")
plt.xlabel("Kelas")
plt.ylabel("Proline")

plt.show()

corr = df[wine.feature_names].corr()

plt.figure(figsize=(10,8))

plt.imshow(corr)

plt.colorbar()

plt.title("Korelasi Antar Fitur")

plt.xticks(
    range(len(wine.feature_names)),
    wine.feature_names,
    rotation=90
)

plt.yticks(
    range(len(wine.feature_names)),
    wine.feature_names
)

plt.show()
```
